```python
# exercicio 1 b)
def ordenada(lista):
    res= True
    if len (lista) > 1:
        if lista[0]> lista[1]:
            res= False 
        else: 
            res = ordenada(lista[1:])
    return res

print(ordenada([2,5,4,3,8]))
print(ordenada([1,2,3,4]))
```

```python
# exercicio 1 c)

def existeElemento(lista,n):
    if lista == []:
        return False
    if lista [0]==n: 
        return True
    return existeElemento(lista[1:],n)

print(existeElemento([1,2,3,4],5))
print(existeElemento([1,2,3,4],2))
```

```python
# exercicio 2 a)

def maiorDivisor(N):
    res = {}
    n = 1

    while n <= N:
        d = 1
        maior = 1

        while d <= 9:
            if n % d == 0:
                maior = d
            d = d + 1

        res[n] = maior
        n = n + 1

    return res
print(maiorDivisor(20))
```

```python
# exercicio 2 b)

def mmc(a,b):
    m=a 
    while m % b != 0:
        m= m + a 
    return m

print(mmc(12, 16))
```

```python
# exercicio 2 c)

fahrenheit = {'t1':-30, 't2':-20, 't3':-10, 't4':0}

def f2c (fahrenheit):
    cel ={}
    for chave in fahrenheit:
        f = fahrenheit [chave]
        c = (f - 32) / 1.8
        cel[chave] = (f,c)
    return cel
print(f2c(fahrenheit))
```

```python
# exercicio 2 d)

def reduxInt(n):
    while n>=10: 
        soma= 0 
        for d in str(n):
            soma= soma + int(d)
        n= soma
    return n 

print(reduxInt(998))
```

```python
# exercicio 3 

# Hospital = [Unidade Hospitalar]
# Unidade Hospitalar = (nome_unidade, [Medico], [Camas])
# Medico = (id, nome)
# Cama = (id, ocupado)
# nome_unidade = String
# nome_medico = String
# id = Int
# ocupado = Bool

hospital = [
 (
 "Unidade de Cardiologia",
 [ (1, "Dr. João"), (2, "Dra. Maria")],
 [(101, False), (102, True), (103, False), (104, False)]
 ),
 (
 "Unidade de Pediatria",
 [ (3, "Dr. Pedro"), (4, "Dra. Clara") ],
 [ (201, True), (202, True), (203, True)]
 )
]

#exercicio a)

def listar_unidades(hospital):
    nomes = []
    for unidade in hospital:
        nomes.append(unidade[0])
    return nomes

print(listar_unidades(hospital))

```

```python
# 3b)
def camas_disponiveis(hospital):
    res=0
    for unidade in hospital:
        camas = unidade[2]
        for cama in camas: 
            if cama[1] == False:
                res = res + 1
    return res

print(camas_disponiveis(hospital))
```

```python
# 3c)

def listar_medicos_unidade (hospital,nome_unidade):
    for unidade in hospital:
        if unidade[0] == nome_unidade:
            medicos = []
            for medico in unidade[1]:
                medicos.append (medico)
            return medicos
    return []

print(listar_medicos_unidade(hospital, "Unidade de Cardiologia"))
```

```python
# 3d)
def existe_medico(hospital, nome_unidade, nome_medico):
    for unidade in hospital:
        if unidade [0] == nome_unidade:
            for medico in unidade[1]:
                if medico [1] == nome_medico:
                    return True
            return False
    return False
print(existe_medico(hospital, "Unidade de Cardiologia", "Dra.Maria"))
```

```python
# 3 e)

def adicionar_medico(hospital, nome_unidade, nome_medico):
    for unidade in hospital:
        if unidade[0] == nome_unidade:

            for medico in unidade[1]:
                if medico[1] == nome_medico:
                    return hospital  
            novo_id = len(unidade[1]) + 1  
            unidade[1].append((novo_id, nome_medico))
            return hospital
    return hospital

print(adicionar_medico(hospital, "Unidade de Cardiologia", "Dra.Paula"))
```

```python
# 3f)

def ocupar_cama(hospital,nome_unidade,id_cama):
    for i in range (len(hospital)):
        unidade=hospital[i]
        if unidade[0] == nome_unidade:
            camas= unidade[2]
            novas_camas=[]
            for cama in camas:
                if cama[0] == id_cama and cama[1]== False:
                    novas_camas.append((cama[0],True))
                else:
                    novas_camas.append(cama)
            hospital[i]= (unidade[0],unidade[1],novas_camas)
            return hospital
        return hospital

    
print(ocupar_cama(hospital, "Unidade de Cardiologia", 101))

```

```python
# 3g)

def guardar_hospital(hospital, nome_ficheiro):
    f=open(nome_ficheiro,"w")
    f.write(str(hospital))
    f.close()
```

```python
# 3h)

def carregar_hospital(nome_ficheiro):
    f=open(nome_ficheiro,"r")
    hospital= eval(f.read())
    f.close()
    return hospital
```

```python
#3i)
def camas_ocupadas(hospital):
    total=0
    ocupadas=0

    for unidade in hospital:
        for cama in unidade[2]:
            total= total + 1
            if cama[1]==True:
                ocupadas = ocupadas + 1
    if total == 0:
        return 0
    return (ocupadas/total)*100
```
